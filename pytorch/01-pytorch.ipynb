```python
import torch
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
print(torch.__version__)
```

```python
# Tensors
# scalar
scalar = torch.tensor(7)
print(scalar)
print(scalar.ndim)
print(scalar.item())
print(scalar.shape)

# vector
vector = torch.tensor([3,4,5])
print(vector)
print(vector.ndim)
print(vector.shape)
print(vector.dtype)
print(vector.device)
# similar stuff goes with matrix and tensors...
```

```python
# Random Tensors
tensor = torch.rand(3,4,4)*4
print(tensor)
print(tensor.ndim)
print(tensor.shape)
# all the usual numpy stuff...
```

```python
img = torch.rand(size=(224, 224, 3)) # sinitializing, ame thing
print(img.shape)
print(img.ndim)
print(img.dtype)
# print(img[0:2][0:2][:5])
```

```python
# other ways of initializing
print(torch.zeros(2,3,2))
print(torch.ones(2,3))
print(torch.zeros_like(torch.arange(0,5))) # make zeros tensor with same shape
```

```python
# range
torch.arange(0,4) # similar to python range()
```
