```python
import numpy as np
from network import Network
```

```python
net = Network(2,2,1)
```

```python
inp = np.array([
	[0,0],
	[0,1],
	[1,0],
	[1,1]
])
output = np.array([[0],[1],[1],[0]])
```

```python
dwB = net.backprop(inp[0], output[0])
dwF = net.finite_diff(inp[0:1], output[0:1], eps=1e-3)
```

```python
# comparing finite diff and backprop
for i in range(len(net.layers)):
    print(np.max(np.abs(
        dwB[i].weights - dwF[i].weights
    )))
```

```python
# compare biases
for i in range(len(net.layers)):
	print(np.max(np.abs(dwB[i].biases - dwF[i].biases)))
```
